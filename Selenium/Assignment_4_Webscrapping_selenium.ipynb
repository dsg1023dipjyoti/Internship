```python
Q2:Scrape the details team India’s international fixtures from bcci.tv.   
Url = https://www.bcci.tv/.   
You need to find following details:   
A) Series   
B) Place   
C) Date   
D) Time  
```

```python
driver= webdriver.Chrome()
driver.get('https://www.bcci.tv/')
```

```python
series_name = driver.find_elements(By.XPATH,'//h5[@class="match-tournament-name ng-binding"]')

for i in series_name:
    series_name=i.text
    series.append(series_name)
```

```python
len(series)
```

```python
dates = driver.find_elements(By.XPATH,'//div[@class="match-dates ng-binding"]')

for i in dates:
    dates=i.text
    date.append(dates)
    
places = driver.find_elements(By.XPATH,'//span[@class="ng-binding"]')

for i in places:
    places=i.text
    place.append(places)
    
times= driver.find_elements(By.XPATH,'//div[@class="match-time no-margin ng-binding"]')
for i in times:
    times=i.text
    time.append(times)
    
```

```python
data= pd.DataFrame({'Series_Name':series,'Date':date,'Place':place,'Time':time})
```

```python
Q8:Details of Datasets from UCI machine learning repositories.   
Url = https://archive.ics.uci.edu/  You 
have to find the following details:   
A) Dataset name   
B) Data type   
C) Task   
D) Attribute type   
E) No of instances   
F) No of attribute G) Year  
```

```python
driver = webdriver.Chrome()
driver.get('https://archive.ics.uci.edu/')
```

```python
data_set = driver.find_element(By.XPATH,'//a[@class="btn-primary btn"]')
data_set.click()
time.sleep(2)
```

```python
data_url=[]
start =0
end =61
for page in range (start,end):
    url=driver.find_elements(By.XPATH,'//a[@class="link-hover link text-xl font-semibold"]')
    for i in url:
        data_url.append(i.get_attribute('href'))
    next_button=driver.find_element(By.XPATH,'/html/body/div/div[1]/div[1]/main/div/div[2]/div[3]/div/button[2]')
    next_button.click()
    time.sleep(2)
```

```python
len(data_url)
```

```python
data_set_name=[]
data_type=[]
task=[]
atri_type=[]
no_of_instance=[]
no_of_attribute=[]
year=[]
```

```python
for url in data_url:
    driver.get(url)
    time.sleep(2)
    
    try:
        datas=driver.find_element(By.XPATH,'//h1[@class="text-3xl font-semibold text-primary-content"]')
        data_set_name.append(datas.text)
    except NoSuchElementException:
        data_set_name.append('-')
    try:
        d_type=driver.find_element(By.XPATH,'//p[@class="text-md"]')
        data_type.append(d_type.text)
    except NoSuchElementException:
        data_type.append('-')
    try:
        tsk=driver.find_element(By.XPATH,'/html/body/div/div[1]/div[1]/main/div/div[1]/div[1]/div[2]/div[2]/div[3]/p')
        task.append(tsk.text)
    except NoSuchElementException:
        task.append('-')
    try:
        attribute=driver.find_element(By.XPATH,'/html/body/div/div[1]/div[1]/main/div/div[1]/div[1]/div[2]/div[2]/div[4]/p')
        atri_type.append(attribute.text)
    except NoSuchElementException:
        atri_type.append('-')
    try:
        ins=driver.find_element(By.XPATH,'/html/body/div/div[1]/div[1]/main/div/div[1]/div[1]/div[2]/div[2]/div[5]/p')
        no_of_instance.append(ins.text)
    except NoSuchElementException:
        no_of_instance.append('-')
    try:
        atr=driver.find_element(By.XPATH,'/html/body/div/div[1]/div[1]/main/div/div[1]/div[1]/div[2]/div[2]/div[6]/p')
        no_of_attribute.append(atr.text)
    except NoSuchElementException:
        no_of_attribute.append('-')
    try:
        yr=driver.find_element(By.XPATH,'/html/body/div/div[1]/div[1]/main/div/div[1]/div[1]/div[1]/div[2]/h2')
        year.append(yr.text)
    except NoSuchElementException:
        year.append('-')
```

```python
data_set = pd.DataFrame({'Dataset_name':data_set_name,'Data_Type':data_type,'Task':task,'Attribute_type':atri_type,
                          'No_of_instances':no_of_instance,'No_of_attribute':no_of_attribute})

data_set
```
